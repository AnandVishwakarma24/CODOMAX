# Module 3 — Data Analysis & Visualization

**Topics:** Pandas, NumPy, data cleaning, analysis, and Matplotlib visualization.

## Objectives
- Load and inspect a dataset
- Clean and prepare data
- Perform basic statistical analysis
- Create bar, line, and pie charts


```python
# 1. Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
# 2. Load the dataset
uploaded = files.upload()
df = pd.read_csv("sales_dataset.csv")
df.head()
```

```python
# 3. Explore the dataset
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
print("\nBasic statistics:")
display(df.describe())
```

```python
# 4. Check for missing and duplicate values
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# 5. Basic data cleaning
df["Date"] = pd.to_datetime(df["Date"])
df = df.drop_duplicates()
df["Sales"] = df["Quantity"] * df["Unit_Price"]

print("Cleaned dataset:")
display(df.head())
```

```python
# 6. Basic analysis
print("Total sales:", df["Sales"].sum())
print("Total quantity sold:", df["Quantity"].sum())
print("Average sales per transaction:", round(df["Sales"].mean(), 2))

sales_by_product = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)
display(sales_by_product)
```

```python
# 7. Sales by region
sales_by_region = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
display(sales_by_region)
```

```python
# 8. Bar chart — Sales by Product
sales_by_product.plot(kind="bar", figsize=(8,5))
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 9. Line chart — Monthly Sales
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales.index = monthly_sales.index.astype(str)

monthly_sales.plot(kind="line", marker="o", figsize=(9,5))
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 10. Pie chart — Sales by Region
sales_by_region.plot(kind="pie", autopct="%1.1f%%", figsize=(7,7))
plt.title("Sales Distribution by Region")
plt.ylabel("")
plt.show()
```

## Conclusion

From this analysis, we can identify:
- The highest-selling products
- Sales trends over time
- Regional sales distribution
- Basic statistics of the dataset
